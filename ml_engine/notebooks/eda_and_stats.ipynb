```python
import pandas as pd
import numpy as np

# 1. Filter dataset to keep Cured and Default cohorts
filtered_df = df[df['Treatment_Outcome'].isin(['Cured', 'Default'])].copy()

# 2. Define risk categorization logic safely
def assign_risk_category(row):
    outcome = row['Treatment_Outcome']
    days = row['Days_To_Interruption']
    
    if outcome == 'Cured':
        return 0  # Low Risk
    elif outcome == 'Default':
        if pd.isna(days):
            return 1  # Fallback to Medium Risk if time is missing
        elif float(days) <= 60:
            return 2  # High Risk (Intensive Phase Dropout)
        else:
            return 1  # Medium Risk (Continuation Phase Dropout)
    return np.nan

# 3. Apply categorization and cast strictly to integer
filtered_df['risk_tier'] = filtered_df.apply(assign_risk_category, axis=1)
filtered_df = filtered_df.dropna(subset=['risk_tier']).copy()
filtered_df['risk_tier'] = filtered_df['risk_tier'].astype(int)

# 4. Map readable labels directly
label_map = {
    0: '0 (Low Risk - Cured)',
    1: '1 (Medium Risk - Late Default)',
    2: '2 (High Risk - Early Default)'
}
filtered_df['risk_label'] = filtered_df['risk_tier'].map(label_map)

# 5. Build summary table safely
counts = filtered_df['risk_label'].value_counts().sort_index()
percentages = (filtered_df['risk_label'].value_counts(normalize=True).sort_index() * 100).round(2)

summary_df = pd.DataFrame({'Patient_Count': counts, 'Percentage': percentages})

print("================ FINAL TARGET CLASS DISTRIBUTION ================")
print(summary_df)
print(f"\nTotal Analyzed Patients: {len(filtered_df)}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import classification_report, roc_auc_score, mean_absolute_error, r2_score, root_mean_squared_error
from xgboost import XGBClassifier, XGBRegressor

# ------------------------------------------------------------------
# 1. DATA PREPARATION
# ------------------------------------------------------------------
# Exclude data leakage columns
leakage_cols = ['Treatment_Outcome', 'Days_To_Interruption', 'risk_tier', 'risk_label']
feature_cols = [c for c in filtered_df.columns if c not in leakage_cols]

X = filtered_df[feature_cols].copy()
y_tier = filtered_df['risk_tier']                     # Target 1: Risk Tier (0, 1, 2)

# Identify numerical vs categorical columns
num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()

# Define shared preprocessors
num_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', num_transformer, num_cols),
    ('cat', cat_transformer, cat_cols)
])

# ------------------------------------------------------------------
# 2. STAGE 1: RISK TIER CLASSIFIER (Low vs Medium vs High)
# ------------------------------------------------------------------
print("================ STAGE 1: TRAINING RISK CLASSIFIER ================")

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X, y_tier, test_size=0.20, random_state=42, stratify=y_tier
)

clf_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', XGBClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=5,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric='mlogloss'
    ))
])

clf_pipeline.fit(X_train_c, y_train_c)

# Evaluate Classifier
y_pred_c = clf_pipeline.predict(X_test_c)
y_proba_c = clf_pipeline.predict_proba(X_test_c)

print("\n--- Classification Report (Risk Tiers) ---")
print(classification_report(y_test_c, y_pred_c, target_names=['Low Risk (0)', 'Medium Risk (1)', 'High Risk (2)'], digits=4))
macro_auc = roc_auc_score(y_test_c, y_proba_c, multi_class='ovr', average='macro')
print(f"🎯 Classifier Macro ROC-AUC: {macro_auc:.4f}")

# ------------------------------------------------------------------
# 3. STAGE 2: DEFAULT TIME REGRESSOR (Defaulters Only)
# ------------------------------------------------------------------
print("\n================ STAGE 2: TRAINING DEFAULT TIME REGRESSOR ================")

# Filter training data strictly to Defaulters (where time_stopped is known)
defaulter_mask = (filtered_df['Treatment_Outcome'] == 'Default') & (filtered_df['Days_To_Interruption'].notnull())
X_defaulters = filtered_df.loc[defaulter_mask, feature_cols]
y_time = filtered_df.loc[defaulter_mask, 'Days_To_Interruption']

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_defaulters, y_time, test_size=0.20, random_state=42
)

reg_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', XGBRegressor(
        n_estimators=150,
        learning_rate=0.03,
        max_depth=4,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    ))
])

reg_pipeline.fit(X_train_r, y_train_r)

# Evaluate Regressor
y_pred_r = reg_pipeline.predict(X_test_r)

mae = mean_absolute_error(y_test_r, y_pred_r)
rmse = root_mean_squared_error(y_test_r, y_pred_r)
r2 = r2_score(y_test_r, y_pred_r)

print(f"• Mean Absolute Error (MAE) : {mae:.2f} days")
print(f"• Root Mean Sq Error (RMSE) : {rmse:.2f} days")
print(f"• R-squared (R²) Score      : {r2:.4f}")

# ------------------------------------------------------------------
# 4. COMBINED INFERENCE FUNCTION
# ------------------------------------------------------------------
def predict_patient_outcome(patient_data_df):
    """
    Accepts a DataFrame of single/multiple patient features.
    Returns predicted risk tier, probability distribution, and estimated days to default.
    """
    risk_tier = clf_pipeline.predict(patient_data_df)[0]
    probabilities = clf_pipeline.predict_proba(patient_data_df)[0]
    
    tier_names = {0: 'Low Risk (Cured)', 1: 'Medium Risk (Late Default)', 2: 'High Risk (Early Default)'}
    
    result = {
        'predicted_risk_tier': tier_names[risk_tier],
        'confidence_scores': {
            'low_risk_prob': round(float(probabilities[0]), 4),
            'medium_risk_prob': round(float(probabilities[1]), 4),
            'high_risk_prob': round(float(probabilities[2]), 4)
        }
    }
    
    # If predicted as Medium or High Risk, estimate time of default
    if risk_tier in [1, 2]:
        est_days = reg_pipeline.predict(patient_data_df)[0]
        result['estimated_days_to_default'] = round(float(est_days), 1)
    else:
        result['estimated_days_to_default'] = None  # Expected to complete treatment
        
    return result

# Test inference on a sample patient from test set
sample_patient = X_test_c.iloc[[0]]
print("\n================ SAMPLE PREDICTION OUTPUT ================")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import classification_report, roc_auc_score, mean_absolute_error, r2_score, root_mean_squared_error
from sklearn.utils.class_weight import compute_sample_weight
from xgboost import XGBClassifier, XGBRegressor

# ------------------------------------------------------------------
# 1. PREPROCESSING PIPELINE
# ------------------------------------------------------------------
leakage_cols = ['Treatment_Outcome', 'Days_To_Interruption', 'risk_tier', 'risk_label']
feature_cols = [c for c in filtered_df.columns if c not in leakage_cols]

X = filtered_df[feature_cols].copy()
y_tier = filtered_df['risk_tier']

num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()

preprocessor = ColumnTransformer(transformers=[
    ('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())]), num_cols),
    ('cat', Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))]), cat_cols)
])

# ------------------------------------------------------------------
# 2. STAGE 1: BALANCED RISK CLASSIFIER
# ------------------------------------------------------------------
print("================ STAGE 1: BALANCED RISK CLASSIFIER ================")

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X, y_tier, test_size=0.20, random_state=42, stratify=y_tier
)

# Compute sample weights to fix 70% class imbalance
sample_weights_train = compute_sample_weight('balanced', y_train_c)

clf_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', XGBClassifier(
        n_estimators=250,
        learning_rate=0.03,
        max_depth=4,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric='mlogloss'
    ))
])

# Fit with balanced sample weights
clf_pipeline.fit(X_train_c, y_train_c, classifier__sample_weight=sample_weights_train)

y_pred_c = clf_pipeline.predict(X_test_c)
y_proba_c = clf_pipeline.predict_proba(X_test_c)

print("\n--- Balanced Classification Report ---")
print(classification_report(y_test_c, y_pred_c, target_names=['Low Risk (0)', 'Medium Risk (1)', 'High Risk (2)'], digits=4))
print(f"🎯 Macro ROC-AUC: {roc_auc_score(y_test_c, y_proba_c, multi_class='ovr', average='macro'):.4f}")

# ------------------------------------------------------------------
# 3. STAGE 2: LOG-TRANSFORMED DEFAULT TIME REGRESSOR
# ------------------------------------------------------------------
print("\n================ STAGE 2: LOG-TRANSFORMED REGRESSOR ================")

defaulter_mask = (filtered_df['Treatment_Outcome'] == 'Default') & (filtered_df['Days_To_Interruption'].notnull())
X_defaulters = filtered_df.loc[defaulter_mask, feature_cols]
y_time = filtered_df.loc[defaulter_mask, 'Days_To_Interruption']

# LOG TRANSFORM: Normalize right-skewed target
y_time_log = np.log1p(y_time)

X_train_r, X_test_r, y_train_r_log, y_test_r_log = train_test_split(
    X_defaulters, y_time_log, test_size=0.20, random_state=42
)

reg_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', XGBRegressor(
        n_estimators=200,
        learning_rate=0.02,
        max_depth=3,
        subsample=0.7,
        colsample_bytree=0.7,
        random_state=42
    ))
])

reg_pipeline.fit(X_train_r, y_train_r_log)

# Predict in log scale and convert back using expm1
y_pred_r_log = reg_pipeline.predict(X_test_r)
y_pred_r = np.expm1(y_pred_r_log)
y_test_r_original = np.expm1(y_test_r_log)

mae = mean_absolute_error(y_test_r_original, y_pred_r)
rmse = root_mean_squared_error(y_test_r_original, y_pred_r)
r2 = r2_score(y_test_r_original, y_pred_r)

print(f"• Mean Absolute Error (MAE) : {mae:.2f} days")
print(f"• Root Mean Sq Error (RMSE) : {rmse:.2f} days")
print(f"• R-squared (R²) Score      : {r2:.4f}")
```

```python
from sklearn.metrics import classification_report, roc_auc_score
from xgboost import XGBClassifier

# ------------------------------------------------------------------
# STEP 1: DEFINE TARGETS FOR HIERARCHICAL PIPELINE
# ------------------------------------------------------------------
# Target A: 0 = Cured, 1 = Defaulter (Tiers 1 & 2)
y_binary_defaulter = (filtered_df['risk_tier'] > 0).astype(int)

X_train_h, X_test_h, y_train_a, y_test_a = train_test_split(
    X, y_binary_defaulter, test_size=0.20, random_state=42, stratify=y_binary_defaulter
)

# ------------------------------------------------------------------
# MODEL A: ADHERENT VS. NON-ADHERENT
# ------------------------------------------------------------------
model_a = Pipeline([
    ('preprocessor', preprocessor),
    ('clf', XGBClassifier(n_estimators=200, max_depth=4, scale_pos_weight=2.3, random_state=42))
])
model_a.fit(X_train_h, y_train_a)

preds_a = model_a.predict(X_test_h)
print("=== MODEL A: Cured vs. Defaulter ===")
print(classification_report(y_test_a, preds_a, target_names=['Cured', 'Defaulter']))

# ------------------------------------------------------------------
# MODEL B: EARLY VS. LATE DEFAULT (DEFAULTERS ONLY)
# ------------------------------------------------------------------
defaulters_only = filtered_df[filtered_df['risk_tier'] > 0].copy()
X_def = defaulters_only[feature_cols]
y_def = (defaulters_only['risk_tier'] == 2).astype(int)  # 1 = Early (Tier 2), 0 = Late (Tier 1)

X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_def, y_def, test_size=0.20, random_state=42, stratify=y_def
)

model_b = Pipeline([
    ('preprocessor', preprocessor),
    ('clf', XGBClassifier(n_estimators=150, max_depth=3, learning_rate=0.03, random_state=42))
])
model_b.fit(X_train_b, y_train_b)

preds_b = model_b.predict(X_test_b)
print("\n=== MODEL B: Late Default (0) vs. Early Default (1) ===")
print(classification_report(y_test_b, preds_b, target_names=['Late Default', 'Early Default']))
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import classification_report, roc_auc_score, precision_recall_curve
from xgboost import XGBClassifier

# 1. Train Model B with explicit scale_pos_weight
pos_weight = (y_train_b == 0).sum() / (y_train_b == 1).sum()  # Ratio of Late to Early

model_b_balanced = Pipeline([
    ('preprocessor', preprocessor),
    ('clf', XGBClassifier(
        n_estimators=150,
        max_depth=3,
        learning_rate=0.03,
        scale_pos_weight=pos_weight,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    ))
])

model_b_balanced.fit(X_train_b, y_train_b)

# 2. Get predicted probabilities for Class 1 (Early Default)
y_probs_b = model_b_balanced.predict_proba(X_test_b)[:, 1]

# 3. Find optimal decision threshold to maximize Macro F1
best_threshold = 0.5
best_macro_f1 = 0.0

for threshold in np.arange(0.2, 0.8, 0.02):
    preds = (y_probs_b >= threshold).astype(int)
    report = classification_report(y_test_b, preds, output_dict=True)
    macro_f1 = report['macro avg']['f1-score']
    
    if macro_f1 > best_macro_f1:
        best_macro_f1 = macro_f1
        best_threshold = threshold

print(f"🎯 Optimal Decision Threshold: {best_threshold:.2f} (Macro F1: {best_macro_f1:.4f})")

# Final predictions with optimal threshold
y_preds_opt = (y_probs_b >= best_threshold).astype(int)
print("\n=== OPTIMIZED MODEL B (Late vs. Early Default) ===")
print(classification_report(y_test_b, y_preds_opt, target_names=['Late Default (1)', 'Early Default (2)'], digits=4))
print(f"ROC-AUC: {roc_auc_score(y_test_b, y_probs_b):.4f}")
```

```python
import os
import joblib
import pandas as pd
import numpy as np

# 1. Create directory for serialized model artifacts
models_dir = "../models"
os.makedirs(models_dir, exist_ok=True)

# 2. Save Model Artifacts & Preprocessor
joblib.dump(preprocessor, f"{models_dir}/tb_preprocessor.joblib")
joblib.dump(model_a, f"{models_dir}/tb_model_a_cured_vs_defaulter.joblib")
joblib.dump(model_b, f"{models_dir}/tb_model_b_early_vs_late.joblib")

print(f"✅ Model artifacts successfully serialized and saved to '{models_dir}/'")

# 3. Unified Production Inference Wrapper
def predict_tb_adherence_risk(patient_df):
    """
    Unified Hierarchical Inference Engine:
    - Step 1: Predict Cured vs Defaulter (Model A)
    - Step 2: If Defaulter, predict Early (Intensive) vs Late (Continuation) Default (Model B)
    """
    # Predict Model A (0 = Cured, 1 = Defaulter)
    pred_a = model_a.predict(patient_df)
    prob_a = model_a.predict_proba(patient_df)[:, 1] # Probability of default
    
    results = []
    
    for idx, (is_defaulter, default_prob) in enumerate(zip(pred_a, prob_a)):
        if is_defaulter == 0:
            tier_code = 0
            tier_name = "Low Risk (Expected Completion)"
            interruption_window = "N/A (Treatment Completion)"
            clinical_action = "Standard Monthly DOTS Monitoring"
            early_default_prob = round(float(1 - default_prob), 4)
        else:
            # Pass patient to Model B for timing stratification
            patient_single = patient_df.iloc[[idx]]
            pred_b = model_b.predict(patient_single)[0]
            prob_b = model_b.predict_proba(patient_single)[0] # [Late, Early]
            
            if pred_b == 1:
                tier_code = 2
                tier_name = "High Risk (Early / Intensive Phase Default)"
                interruption_window = "Days 0–60 (Intensive Phase)"
                clinical_action = "Immediate DOTS Enabler, Weekly Follow-ups, Intensive Counseling"
                early_default_prob = round(float(prob_b[1]), 4)
            else:
                tier_code = 1
                tier_name = "Medium Risk (Late / Continuation Phase Default)"
                interruption_window = "Days 61–180 (Continuation Phase)"
                clinical_action = "Bi-weekly Digital Reminders, Side-effect Management Check"
                early_default_prob = round(float(prob_b[1]), 4)

        results.append({
            "risk_tier_code": tier_code,
            "risk_tier_name": tier_name,
            "overall_default_probability": round(float(default_prob), 4),
            "estimated_interruption_window": interruption_window,
            "recommended_clinical_action": clinical_action
        })
        
    return pd.DataFrame(results)
```
